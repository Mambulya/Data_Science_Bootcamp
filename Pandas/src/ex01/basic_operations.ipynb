```python
import pandas as pd

src_file = "../../feed-views.log"
```

## 1) Create a dataframe views with two columns

```python
df = pd.read_csv(src_file, sep='\t', header=None)
df.columns=["datetime", "user"]
```

Проверка типов:

```python
df.dtypes
```

Изменение типа datetime на datetime64:

```python
df["datetime"] = df["datetime"].astype("datetime64")
```

```python
df.dtypes
```

Разбиение колонки datetime на year, month, day, hour, minute, and second:

```python
df["year"] = df["datetime"].dt.year
df["month"] = df["datetime"].dt.month
df["day"] = df["datetime"].dt.day
df["hour"] = df["datetime"].dt.hour
df["minute"] = df["datetime"].dt.minute
df["second"] = df["datetime"].dt.second
```

```python
df
```

```python
df.info()
```

## 2) Create the new column daytime

Зададим колонку user ввиде ключа. Разделим колонку hour на иданные интервалы:

0 – 3.59 night, 4 – 6.59 early morning, 7 – 10.59 morning, 11 – 16.59 afternoon, 17 – 19.59 early evening, 20 – 23.59 evening

```python
df["daytime"] = pd.cut(df["hour"], bins=[0, 4, 7, 11, 17, 20, 24], right=False, labels=["night", "early morning", "morning", "afternoon", "early evening", "evening"])
df.set_index("user", inplace=True)
```

## 6) Show the 3 earliest hours in the morning and the corresponding usernames and the 3 latest hours and the usernames using nsmallest() and nlargest()

3 earlist hours:

```python
morning_df = df[df["daytime"] == "morning"]
res = morning_df.nsmallest(n=3, columns="hour")
final = pd.DataFrame({"user":res.index, "hour":res["hour"]}) # без reset_index()
print(final.to_string(index=False))
```

3 latest ones:

```python
morning_df = df[df["daytime"] == "morning"]
res = morning_df.nlargest(n=3, columns="hour")
final = pd.DataFrame({"user":res.index, "hour":res["hour"]}) # без reset_index()
print(final.to_string(index=False))
```

### Find out what the most popular interval for visiting the page is (interquartile range)

```python
Q1 = df.describe()["hour"]["25%"]  # 25% часов находятся ниже этого часа Q1
Q3 = df.describe()["hour"]["75%"]  
iqr = Q3 - Q1
print(f"Q1 = {Q1}   Q3 = {Q3}")
print(f"iqr = {iqr}")
```
