 Pandas gives the tools not only to perform descriptive analysis and understand data better but to _preprocess_ it too.

```python
import pandas as pd
src_file = "../../auto.json"
```

## 1) Create a dataframe and make ID the index column

```python
df = pd.read_csv(src_file, sep=',', orient="records")
df.set_index("ID", inplace=True)
display(df)
```

```python
df["Refund"].mean()
```

## 2) Count the number of observations using the method count()

```python
df.count()
```

## 3) Drop the duplicates, taking into account only the following columns: CarNumber, Make_n_model, Fines

* between the two equal observations, you need to choose the last
* check the number of observations again

```python
df.drop_duplicates(subset=["CarNumber", "Make_n_model", "Fines"], keep="last", inplace=True)
display(df)
print(df.count())
```

## 4) Work with missing values

* check how many values are missing from each column

```python
df.isnull().sum()
```

* drop all the columns with over 500 missing values using the argument thresh, check how many missing values are in each column

```python
df.dropna(axis="columns", thresh=len(df) - 500, inplace=True)
df.isnull().sum()
```

* replace all the missing values in the Refund column with the previous value in that column for that cell, use the _argument method_, check how many values are missing from each column

```python
# покажем одну из строчке в Refund до изменений
df.loc[[146,147]]
```

```python
# видно, что ячейка поменялась
df["Refund"].fillna(method="ffill", inplace=True)
display(df.loc[[146, 147]])
print(df.isnull().sum())
```

* replace all the missing values in the Fines column with the mean value of this column (exclude NA/null values when computing the mean value), check how many values are missing from each column

```python
fines_mean = df["Fines"].mean()
df["Fines"].fillna(fines_mean, inplace=True)
print(df.isnull().sum())
```

## 5) Split and parse the make and model

* use the method _apply_ both for splitting and for extracting the values to the new columns Make and Model

* drop the column Make_n_model

```python
def get_make(row):
    parsed = row["Make_n_model"].split()
    return parsed[0]

def get_model(row):
    parsed = row["Make_n_model"].split()
    if len(parsed) == 1:
        return None
    else:
        return parsed[1]

df["Make"] = df.apply(get_make, axis=1)
df["Model"] = df.apply(get_model, axis=1)
df.drop("Make_n_model", axis="columns", inplace=True)
display(df)
```

* save the dataframe in the JSON file auto.json in the format below

```python
file_to_save = "auto.json"
df.to_json(file_to_save, double_precision=2)
```
