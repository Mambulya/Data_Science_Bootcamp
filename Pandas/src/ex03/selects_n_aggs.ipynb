## 1) Load the JSON file that you created in the previous exercise into a dataframe

* set CarNumber as the index column

```python
src_file =  "../ex02/auto.json"
df = pd.read_json(src_file)
df.set_index("CarNumber", inplace=True)
display(df)
```

## 2) Make the following selects

* display the rows only where the __fines__ are more than 2,100

```python
display(df[df["Fines"] > 2100])
```

* display the rows only where the __fines__ are more than 2,100 and the __refund__ equals 2

```python
display(df[(df["Fines"] > 2100) & (df["Refund"] == 2)])
```

* display the rows only where the __models__ are from the list: [’Focus’, ’Corolla’]

```python
display(df[df["Model"].isin(["Focus", "Corolla"])])
```

* display the rows only where the __car number__ is from the list: [’Y7689C197RUS’, ’92928M178RUS’, ’7788KT197RUS’, ’H115YO163RUS’, ’X758HY197RUS’]

```python
car_list = ["Y7689C197RUS", "92928M178RUS", "7788KT197RUS", "H115YO163RUS", "X758HY197RUS"]
display(df[df.index.isin(car_list)])  # колонка CarNumber - индекс
```

## 3) Make the aggregations with the make and the model

* display the median __fines__ grouped by the __make__

```python
display(df.groupby(by="Make")["Fines"].median())
```

* display the median __fines__ grouped by the __make__ and __the model__

```python
display(df.groupby(["Make", "Model"])["Fines"].median())
```

* display the number of __fines__ grouped by the __make__ and the __model__ in order to understand if we can trust the median values

```python
display(df.groupby(["Make", "Model"])["Fines"].count())
```

* display the minimum and the maximum __fines__ grouped by the __make__ and the __model__ in order to better understand the variance

```python
display(df.groupby(["Make", "Model"])["Fines"].min())
```

```python
display(df.groupby(["Make", "Model"])["Fines"].max())
```

или же одной строчкой:

```python
display(df.groupby(by=["Make", "Model"]).aggregate({"Fines":["min", "max"]}))
```

* display the standard deviation of the __fines__ grouped by the __make__ and the __model__ in order to better understand the variance

```python
display(df.groupby(by=["Make", "Model"])["Fines"].std())
```

## 4) Make the aggregations with the car number

* display the __car numbers__ grouped by the __number of the fines__ in _descending order_, we want to find those who most often violated the law

```python
display(df.groupby(by="CarNumber").aggregate({"Fines":["count"]}).sort_values(by=("Fines", "count"), ascending=False))
```

* select from the initial dataframe all the rows corresponding to the top-1 car number, we want to zoom in a little bit

```python
grouped = df.groupby(by="CarNumber").aggregate({"Fines":["count"]}).sort_values(by=("Fines", "count"), ascending=False)
topCar = grouped.index[0]
display(df[df.index == topCar])
```

* display the car numbers grouped by the sum of the __fines__ in _descending order_, we want to find those who paid the most

```python
display(df.groupby(by="CarNumber").aggregate(func={"Fines":["sum"]}).sort_values(by=("Fines", "sum"), ascending=False))
```

* select from the initial dataframe all the rows corresponding to the top-1 car number, we want to zoom in a little bit

```python
g = df.groupby(by="CarNumber").aggregate(func={"Fines":["sum"]}).sort_values(by=("Fines", "sum"), ascending=False)
top_car = g.index[0]
display(df[df.index == top_car])
```

* display a table that answers the question: are there any car numbers that were connected to different models?

```python
res = df.groupby(by=["CarNumber", "Model"]).aggregate(func={"Model":["nunique"]}).sort_values(by=("Model","nunique"), ascending=False)
display(res)
print("Видно, что у всех номеров только 1 модель")
```
