## 2) Iterations

Сalculate __fines/refund*year__ for each row and create a new column with the calculated data and measure the time using the magic command %%timeit in the cell

### 1. Loop

```python
def calculate_rows():
    values = []
    for i in range(len(df)):
        fines = df.iloc[i, 2]   # iloc не считывает индексы
        refund = df.iloc[i, 1]
        year = df.iloc[i, 5]
        values.append(fines/refund*year)
    return values
```

```python
%%timeit 
df["Res Loop"] = calculate_rows()
```

###  2. Iterrows()

```python
def calculate_iter():
    values = []
    for index, row in df.iterrows():
        fines = row["Fines"]
        refund = row["Refund"]
        year = row["Year"]
        values.append(fines/refund*year)
    return values
```

```python
%%timeit 
df["Res Iter"] = calculate_iter()
```

### 3. Apply() and lambda function

```python
%%timeit
df["Res Lambda"] = df.apply(lambda row: row["Fines"] / row["Refund"] * row["Year"], axis=1)  # приметить функцию построчно
```

### 4. Series objects from the dataframe

```python
%%timeit
df["Res Series"] = df["Fines"] / df["Refund"] * df["Year"]
```

### 5. .values

```python
%%timeit
df["Res Values"] = df["Fines"].values / df["Refund"].values * df["Year"].values
```

В итоге по эффективности получилось:
1) apply() + lambda
2) itterrows()
2) loop
3) series + .values

* make a copy() of your initial dataframe into another dataframe optimized

Проанализируем диапазон значений в численных колонках:

```python
for column in df.columns:
    try:
        print(f"|Колонка '{column}'| занчения от {opt_df[column].min()} до {opt_df[column].max()}")
    except (SyntaxError, TypeError):
        print(f"|Колонка '{column}'| не численное значение")
```

В итоге:
1) Для колонки Refund: int64 (который вмещает от/до ± 9 223 372 036 854 775 807) -> uint8 (0;255) (8 байт -> 1 байт)
Результат: -6.3 KB

```python
opt_df["Refund"] = opt_df["Refund"].astype("uint8")
opt_df.info(memory_usage="deep")
```

2) Колонка "Fines" float64 (±1.79e+308) -> float32 (±3.40E+38) (8 байт -> 4 байта)
Результат: - 3.7 KB

```python
opt_df["Fines"] = opt_df["Fines"].astype("float32")
opt_df.info(memory_usage="deep")
```

3) Колонки Make и Model: определим точный тип
Результат: - 8.7 KB

```python
opt_df['Make'] = opt_df['Make'].astype('category')
opt_df['Model'] = opt_df['Model'].astype('category')
opt_df.info(memory_usage="deep")
```

4. Аналогично для колонок Year, Res:

```python
opt_df["Year"] = opt_df["Year"].astype("int16")
opt_df["Res Loop"] = opt_df["Res Loop"].astype("float32")
opt_df["Res Iter"] = opt_df["Res Iter"].astype("float32")
opt_df["Res Lambda"] = opt_df["Res Lambda"].astype("float32")
opt_df["Res Series"] = opt_df["Res Series"].astype("float32")
opt_df["Res Values"] = opt_df["Res Values"].astype("float32")
opt_df.index = opt_df.index.astype("category")
```

```python
df.info(memory_usage="deep")
opt_df.info(memory_usage="deep")
```
