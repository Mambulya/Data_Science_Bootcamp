```python
import pandas as pd

# Создаем простой DataFrame
data = {
    'Имя': ['Алекс', 'Мария', 'Иван', 'Елена'],
    'Возраст': [23, 30, 22, 35],
    'Город': ['Москва', 'Санкт-Петербург', 'Казань', 'Новосибирск']
}

df = pd.DataFrame(data)

# Выводим DataFrame
print("Исходный DataFrame:")
print(df)

# Добавим новый столбец с данными о зарплате
df['Зарплата'] = [50000, 60000, 55000, 70000]

# Выводим обновленный DataFrame
print("\nОбновленный DataFrame с зарплатой:")
print(df)

# Фильтруем данные: выбираем людей старше 25 лет
filtered_df = df[df['Возраст'] > 25]

# Выводим отфильтрованный DataFrame
print("\nЛюди старше 25 лет:")
print(filtered_df)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Создаем массив значений x от 0 до 2π
x = [0,1,2,3]

# Вычисляем значения y как синус от x
y = [0,-1,-2,-3]

# Создаем график
plt.figure(figsize=(10, 5))  # Устанавливаем размер графика
plt.plot(x, y, label='sin(x)', color='blue')  # Строим линию графика

# Добавляем заголовок и метки осей
plt.title('График функции sin(x)')
plt.xlabel('x (радианы)')
plt.ylabel('sin(x)')

# Добавляем сетку
plt.grid(True)

# Добавляем легенду
plt.legend()

# Показываем график
plt.show()
```
